```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
datos = {
    'Habitaciones': [23, 47, 21, 39, 37, 29, 23, 44, 45, 16, 30, 42, 54, 
                     27, 34, 15, 19, 38, 44, 47, 43, 38, 51, 61, 39],
    'Ingreso': [1452, 1361, 1426, 1470, 1456, 1430, 1354, 1442, 1394, 1459, 1399, 1458, 1537, 
                1425, 1445, 1439, 1348, 1450, 1431, 1446, 1485, 1405, 1461, 1490, 1426]
}
```

```python
#cálculo del coeficiente de correlación
df = pd.DataFrame(datos)

correlacion = df['Habitaciones'].corr(df['Ingreso'])
print(f"El coeficiente de correlacion es: {correlacion:.4f}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Habitaciones'], df['Ingreso'], color='blue', label='Días observados')

m,b = np.polyfit(df['Habitaciones'], df['Ingreso'], 1)
plt.plot(df['Habitaciones'], m*df['Habitaciones'] + b, color='orange', label='Línea de tendencia')


plt.title('Correlación: Habitaciones Ocupadas vs Ingreso del Restaurante')
plt.xlabel('Número de Habitaciones Ocupadas')
plt.ylabel('Ingreso del Restaurante ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
