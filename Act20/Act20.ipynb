```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from itertools import permutations
import random
import time

np.random.seed(42)
random.seed(42)

print(" Librerías importadas correctamente")
```

```python
class ColoniaHormigas:    
    def __init__(self, distancias, n_hormigas=10, alpha=1.0, beta=2.0,
                 rho=0.5, Q=100, n_iter=100):
        self.distancias = np.array(distancias, dtype=float)
        self.n_ciudades = len(distancias)
        self.n_hormigas = n_hormigas
        self.alpha = alpha
        self.beta = beta
        self.rho = rho
        self.Q = Q
        self.n_iter = n_iter
        
        # Inicializar feromonas con valor uniforme
        self.feromonas = np.ones((self.n_ciudades, self.n_ciudades))
        
        # Calcular visibilidad (1/distancia), evitar división por cero
        with np.errstate(divide='ignore'):
            self.visibilidad = np.where(
                self.distancias > 0, 1.0 / self.distancias, 0
            )
        
        # Historial de la mejor solución
        self.mejor_ruta = None
        self.mejor_distancia = np.inf
        self.historial_distancias = []

print("✅ __init__ definido")
```

```python
def _calcular_probabilidades(self, ciudad_actual, visitadas):
        no_visitadas = [c for c in range(self.n_ciudades) if c not in visitadas]
        
        if not no_visitadas:
            return [], []
        
        # Numeradores: τ^α × η^β
        numeradores = np.array([
            (self.feromonas[ciudad_actual][j] ** self.alpha) *
            (self.visibilidad[ciudad_actual][j] ** self.beta)
            for j in no_visitadas
        ])
        
        total = numeradores.sum()
        if total == 0:
            probabilidades = np.ones(len(no_visitadas)) / len(no_visitadas)
        else:
            probabilidades = numeradores / total
        
        return no_visitadas, probabilidades

# Agregar método a la clase
ColoniaHormigas._calcular_probabilidades = _calcular_probabilidades
print(" calcular_probabilidades definido")
```

```python
def _construir_ruta(self):
        """Una hormiga construye una ruta completa."""
        ciudad_inicio = random.randint(0, self.n_ciudades - 1)
        ruta = [ciudad_inicio]
        visitadas = {ciudad_inicio}
        
        while len(ruta) < self.n_ciudades:
            ciudad_actual = ruta[-1]
            candidatas, probs = self._calcular_probabilidades(ciudad_actual, visitadas)
            
            if not candidatas:
                break
            
            # Selección probabilística (ruleta)
            siguiente = np.random.choice(candidatas, p=probs)
            ruta.append(siguiente)
            visitadas.add(siguiente)
        
        return ruta

ColoniaHormigas._construir_ruta = _construir_ruta
print("✅ _construir_ruta definido")
```

```python
def _calcular_distancia_ruta(self, ruta):
        """Calcula la distancia total de una ruta (incluyendo regreso al inicio)."""
        distancia = sum(
            self.distancias[ruta[i]][ruta[i+1]]
            for i in range(len(ruta) - 1)
        )
        distancia += self.distancias[ruta[-1]][ruta[0]]  # Regreso al inicio
        return distancia

ColoniaHormigas._calcular_distancia_ruta = _calcular_distancia_ruta
print("✅ _calcular_distancia_ruta definido")
```

```python
def _actualizar_feromonas(self, rutas, distancias_rutas):
        """Actualiza el rastro de feromonas: evaporación + depósito."""
        # Evaporación
        self.feromonas *= (1 - self.rho)
        
        # Depósito por cada hormiga
        for ruta, dist in zip(rutas, distancias_rutas):
            delta = self.Q / dist
            for i in range(len(ruta) - 1):
                self.feromonas[ruta[i]][ruta[i+1]] += delta
                self.feromonas[ruta[i+1]][ruta[i]] += delta  # Simétrico
            # Arco de regreso
            self.feromonas[ruta[-1]][ruta[0]] += delta
            self.feromonas[ruta[0]][ruta[-1]] += delta

ColoniaHormigas._actualizar_feromonas = _actualizar_feromonas
print("✅ _actualizar_feromonas definido")
```
