```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
dt = pd.read_csv("celsius_a_fahrenheit.csv")
dt.head()
```

```python
sns.scatterplot(x="Celsius", y="Fahrenheit", data=dt)

plt.title("Correlacion entre celsius y fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")

plt.show()
```

```python
X_train = dt["Celsius"]
Y_train = dt["Fahrenheit"]
```

```python
print("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(units=1))
```

```python
print("Compilando el programa")
model.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss="mean_squared_error")
```

```python
print("entrenamiento del programa")
epochs_hist = model.fit(X_train, Y_train, epochs=100)
```

```python
print("Evaluando el modelo entrenado")
print("Keys")
print(epochs_hist.history.keys())

plt.plot(epochs_hist.history["loss"])
plt.title("Progreso de perdida durante el entrenamiento del modelo")
plt.xlabel("Epoch o Veces que se itero")
plt.ylabel("Training loss")
plt.legend(["Training Loss"])

plt.show()
```

```python
#nos dice que solo va a recibir un valor númerico(input(shape=1))
capa_entrada = tf.keras.Input(shape=(1,))
#nos dice que solo va a tener una neurona para procesar la informacion
capa_unica = tf.keras.layers.Dense(units=1)
```

```python
# Definimos las capas como variables (Punto 6)
entrada = tf.keras.layers.Input(shape=(1,))
oculta1 = tf.keras.layers.Dense(units=3)
oculta2 = tf.keras.layers.Dense(units=3)
salida = tf.keras.layers.Dense(units=1)

# Las unimos en el modelo secuencial
modelo_complejo = tf.keras.Sequential([
    entrada,
    oculta1,
    oculta2,
    salida
])
```

```python
print(oculta1.get_weights())
print(oculta2.get_weights())
print(salida.get_weights())
```

```python
print("Compilando el programa")
modelo_complejo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss="mean_squared_error")
```

```python
print("entrenamiento del programa")
epochs_hist = model.fit(X_train, Y_train, epochs=100)
```

```python
print("Evaluando el modelo entrenado")
print("Keys")
print(epochs_hist.history.keys())


plt.plot(epochs_hist.history["loss"])
plt.title("Progreso de perdida durante el entrenamiento del modelo")
plt.xlabel("Epoch o veces de iteracion")
plt.ylabel("Training loss")
plt.legend(["Training loss"])
plt.show()
```
