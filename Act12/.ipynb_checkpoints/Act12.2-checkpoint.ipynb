```python
import math 
import numpy as np 
import matplotlib.pyplot as plt 
import logging
```

```python
import tensorflow as tf
import tensorflow_datasets as tfds
```

```python
logger = tf.get_logger()
logger.setLevel(logging.ERROR)
```

```python
logger = tf.get_logger()
logger.setLevel(logging.ERROR)
```

```python
dataset, metadata = tfds.load("mnist", as_supervised=True, with_info=True)
train_dataset,test_dataset = dataset["train"],dataset["test"]
```

```python
class_names = ["cero","uno","dos","tres","cuatro","cinco","seis","siete", "ocho", "nueve"]
```

```python
num_train_examples = metadata.splits["train"].num_examples
num_test_examples = metadata.splits["test"].num_examples
```

```python
def normalize(images,labels):
    images = tf.cast(images,tf.float32)
    images /= 255
    return images, labels
```

```python
train_dataset = train_dataset.map(normalize)
test_dataset = test_dataset.map(normalize)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28,1)), #capa de entrada
    tf.keras.layers.Dense(64, activation=tf.nn.relu), #capa oculta
    tf.keras.layers.Dense(64, activation=tf.nn.relu), #capa oculta
    tf.keras.layers.Dense(10,activation=tf.nn.softmax)
])
    
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
BATCHSIZE = 32
train_dataset = train_dataset.repeat().shuffle(num_train_examples).batch(BATCHSIZE)
test_dataset = test_dataset.batch(BATCHSIZE)
```

```python
model.fit(
    train_dataset,epochs=5,
    steps_per_epoch=math.ceil(num_train_examples/BATCHSIZE)
)
```

```python
from urllib import parse
from http.server import HTTPServer, BaseHTTPRequestHandler
```

```python
class SimpleHTTPRequestHandler(BaseHTTPRequestHandler):
    
    def do_POST(self):
        print("Peticion recibida")
        
        # Leer cuántos bytes vienen en la petición y luego leerlos
        content_length = int(self.headers['Content-Length'])
        data = self.rfile.read(content_length)
        
        # Limpiar: quitar "pixeles=" al inicio y decodificar caracteres especiales
        # (los datos llegan como "pixeles=0.12,0.00,0.98,...")
        data = data.decode().replace('pixeles=', '')
        data = parse.unquote(data)
        
        # Transformar el string de números al formato que espera el modelo:
        # string → array de 784 floats → matriz 28x28 → tensor 1x28x28x1
        arr = np.fromstring(data, np.float32, sep=",")
        arr = arr.reshape(28, 28)
        arr = np.array(arr)
        arr = arr.reshape(1, 28, 28, 1)  # 1=batch, 28x28=imagen, 1=canal de color
        
        # Usar el modelo ya entrenado para predecir
        prediction_values = model.predict(arr, batch_size=1)
        prediction = str(np.argmax(prediction_values))  # El índice con mayor probabilidad
        print("Prediccion final: " + prediction)
        
        # Enviar la respuesta de vuelta al HTML
        self.send_response(200)
        self.send_header("Access-Control-Allow-Origin", "*")  # Evita errores CORS
        self.end_headers()
        self.wfile.write(prediction.encode())
```

```python
# Esta celda bloquea el kernel mientras el servidor escucha
# Para detenerlo: Kernel > Interrupt en el menú de Jupyter
print("Iniciando el servidor...")
server = HTTPServer(('localhost', 8000), SimpleHTTPRequestHandler)
server.serve_forever()
```
