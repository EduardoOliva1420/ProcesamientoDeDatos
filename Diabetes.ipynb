```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
diabetes = load_diabetes(as_frame=True)
df_diabetes = diabetes.frame
df_diabetes.head()
```

```python
# Variable independiente
X = df_diabetes[['bmi']]

# Variable dependiente
y = df_diabetes['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
y_pred = regressor.predict(X_test)
```

```python
print(f"Mean squared error: {mean_squared_error(y_test, y_pred):.2f}")
print(f"Coefficient of determination: {r2_score(y_test, y_pred):.2f}")
```

```python
plt.figure(figsize=(8, 6))
# Datos reales
plt.scatter(X_test, y_test, label="Datos reales")
# Predicciones del modelo
plt.plot(X_test, y_pred, label="Regresión lineal", linewidth=2)
plt.xlabel("BMI (IMC)")
plt.ylabel("Progresión de la enfermedad")
plt.title("Regresión lineal: IMC vs Progresión de la enfermedad")
plt.legend()
plt.grid(True)
plt.show()
```

En este trabajo se implementó un modelo de regresión lineal para analizar la relación entre el índice de masa corporal (IMC) y la progresión de la enfermedad utilizando el dataset diabetes de Scikit-Learn.
Los resultados obtenidos muestran que existe una relación positiva entre el IMC y el avance de la enfermedad, lo cual indica que a mayor IMC, mayor es la progresión estimada.
El modelo obtuvo un error cuadrático medio de 2548.07 y un coeficiente de determinación de 0.47, lo que significa que el IMC explica aproximadamente el 47% de la variabilidad en la progresión de la enfermedad.
